```python
from models.monte_carlo import AmericanMCAnalyzer, EuropeanMCAnalyzer
```

```python
contract = {
    "spot": 100.0,
    "strike": 105.0,
    "time_to_maturity": 1.0,
    "risk_free_rate": 0.03,
    "volatility": 0.25,
    "option_type": "call",
}
```

```python
european_analyzer = EuropeanMCAnalyzer(num_simulations=5000_000, seed=42)
american_analyzer = AmericanMCAnalyzer(
    num_simulations=500_000,
    num_steps=500,
    seed=42,
    regression_type="quadratic",
)
```

```python
result_european = european_analyzer.analyze(contract)
result_american = american_analyzer.analyze(contract)
```

```python
print("European:", result_european)
print("American:", result_american)
```
